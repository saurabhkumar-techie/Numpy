```python
import numpy as np
a = np.array([10,20,30])
b = np.array([1,2,3])
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)#floor division)
```

```python
print(np.array([-7]) // 2)
```

```python
#modulus
x = np.array([10,17,25])
y = np.array([3,5,7])
print(x % y)
```

```python
#power
x = np.array([2 ,3 , 4])
print(x ** 3)
```

```python
#Scalar arithmetic
arr = np.array([10, 20, 30])
print(arr + 5)
print(arr - 5)
print(arr * 2)
print(arr / 10)
```

```python
#Compound operations
arr = np.array([10, 20, 30])
result = (arr + 5) * 2
print(result)
```

```python
#Inplace operations
arr = np.array([10, 20, 30])
arr += 5
print(arr)
```

```python
marks = np.array([65,75,80,85,90])
update_marks = marks + 5
print("original marks: ",marks)
print("updated marks: ",update_marks)
```
