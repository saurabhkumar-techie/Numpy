```python
import numpy as np
print("Numpy imported successfully.")
print("Numpy version: ",np.__version__)
```

```python
#python list
python_list = [10, 20, 30, 40, 50]

#Numpy array
numpy_array = np.array([10, 20, 30, 40, 50])

print("python list: ",python_list)
print("Numpy array: ",numpy_array)

print("list type: ",type(python_list))
print("array type: ",type(numpy_array))
```

```python
numbers = np.array([10, 20, 30, 40, 50])
print("original: ",numbers)
print("add 5: ",numbers + 5)
print("subtract 5: ",numbers - 5)
print("multiply 5: ",numbers * 5)
print("divide 5: ",numbers / 5)
print("power 5: ",numbers ** 5)
print("square: ",numbers * 2)
print("square root: ",np.sqrt(numbers))
#vectorized operation
```
