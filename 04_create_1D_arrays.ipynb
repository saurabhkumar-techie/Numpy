```python
#array
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print(arr)
print("dimensions: ",arr.ndim)
print("shape: ",arr.shape)
```

```python
#creating a integer array
integers = np.array([10, 20, 30, 40, 50])

print(integers)
print("data types: ",integers.dtype)
```

```python
#creating float array
float = np.array([10.50, 20.50, 30.50, 40.50])
print(float)
print("data types: ",float.dtype)
```

```python
#creating string array
name = np.array(["Saurabh", "Aman", "Aditya"])
print(name)
print("data types: ",name.dtype)
```

```python
#creating boolean arrays
status = np.array([True, False, True, True, False])
print(status)
print("data type: ",status.dtype)
```

```python
#mixed data type arrays
arr = np.array([10, 20.5, 30, 40.5])
print(arr)
print(arr.dtype)
```

```python
#explicit data type using dtype

# arr = np.array([1, 2, 3, 4, 5], dtype=float)
# print(arr)
# print(arr.dtype)

arr = np.array([1.5, 2.5, 3.5], dtype=int)
print(arr)
print(arr.dtype)
```

```python
#type conversion using astype()

arr = np.array([10,20,30,40])
float_arr = arr.astype(float)

print("original: ",arr)
print("converted: ",float_arr)

print(arr.dtype)
print(float_arr.dtype)

#float name overwrite
```

```python
import numpy as np

marks = np.array([85,90,78,92,89])

print("marks: ",marks)
print("total marks: ",np.sum(marks))
print("average marks: ",np.mean(marks))
print("highest marks: ",np.max(marks))
print("lowest marks: ",np.min(marks))
```
