```python
#python list
numbers = ([10, 20, 30, 40, 50])
print(numbers)
print(type(numbers))
```

```python
#Numpy array
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])
print(numbers)
print(type(numbers))
```

```python
#list vs array concatenation
numbers = [10,20,30]
print(numbers + [5])
print(type(numbers))
```

```python
#Numpy array
import numpy as np
numbers = np.array([10, 20, 30])
print(numbers +  5)
```

```python
#multiplication comparison
#python list
numbers = [1, 2, 3]
print(numbers * 2)

#Numpy array
arr = np.array([1, 2, 3])
print(arr * 2)
```

```python
#mathematical operation
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("addition: ",arr + 10)
print("subtraction: ",arr - 5)
print("multiplication: ",arr * 2)
print("division: ",arr / 10)
print("square: ",arr ** 2)
```

```python
#performance comparison
import numpy as np
import time

size = 1_000_000

# Python List
python_list = list(range(size))

start = time.perf_counter()

result = [x * 2 for x in python_list]

list_time = time.perf_counter() - start


# NumPy Array
numpy_array = np.arange(size)

start = time.perf_counter()

result = numpy_array * 2

numpy_time = time.perf_counter() - start

print("Python List Time:", list_time)
print("NumPy Array Time:", numpy_time)
print("NumPy Speedup:", list_time / numpy_time)
```

```python
#memory comparison
import sys
import numpy as np

python_list = list(range(1000))
numpy_array = np.arange(1000)

list_memory = sys.getsizeof(python_list)
array_memory = numpy_array.nbytes

print("List container memory:", list_memory, "bytes")
print("NumPy array data memory:", array_memory, "bytes")
```
