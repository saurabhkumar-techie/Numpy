```python
#python list
numbers = ([10, 20, 30, 40, 50])
print(numbers)
print(type(numbers))
```

```python
#Numpy array
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])
print(numbers)
print(type(numbers))
```

```python
#list vs array concatenation
numbers = [10,20,30]
print(numbers + [5])
print(type(numbers))
```

```python
#Numpy array
import numpy as np
numbers = np.array([10, 20, 30])
print(numbers +  5)
```

```python
#multiplication comparison
#python list
numbers = [1, 2, 3]
print(numbers * 2)

#Numpy array
arr = np.array([1, 2, 3])
print(arr * 2)
```

```python
#mathematical operation
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("addition: ",arr + 10)
print("subtraction: ",arr - 5)
print("multiplication: ",arr * 2)
print("division: ",arr / 10)
print("square: ",arr ** 2)
```
