```python
#python list
numbers = ([10, 20, 30, 40, 50])
print(numbers)
print(type(numbers))
```

```python
#Numpy array
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])
print(numbers)
print(type(numbers))
```

```python
#list vs array concatenation
numbers = [10,20,30]
print(numbers + [5])
print(type(numbers))
```

```python
#Numpy array
import numpy as np
numbers = np.array([10, 20, 30])
print(numbers +  5)
```
