```python
import numpy as np
```

```python
#identity matrix using np.eye
identity = np.eye(3)
print(identity)
```

```python
identity = np.eye(3, dtype=int)
print(identity)
```

```python
#np.identity
matrix = np.identity(4, dtype=int)
print(matrix)
```

```python
matrix = np.eye(2, 4, dtype=int)
print(matrix)
```

```python
#Diagonal matrix
diagonal = np.diag([2, 5, 8])
print(diagonal)
```

```python
diagonal = np.diag([10, 20, 30, 40])
print(diagonal)
```

```python
#Extract diagonal values
matrix = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])
result = np.diag(matrix)
print(result)
```

```python
matrix = np.eye(3, dtype=int, k=1)
print(matrix)
```

```python
#matrix operation
A = np.array([
    [2, 3],
    [4, 5]
])

I = np.eye(2, dtype=int)

result = A @ I #@ multiplication operator use in algebra and ml algorithm
print(result)
```
