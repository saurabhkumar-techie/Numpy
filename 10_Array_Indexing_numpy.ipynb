```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])

print(arr[0])
print(arr[2])
print(arr[4])
# print(arr[6]) IndexError: index 6 is out of bounds for axis 0 with size 5
```

```python
#last element access
arr = np.array([10,20,30,40,50])
print(arr[-1])
print(arr[-2])
```

```python
#2D array indexing
matrix = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])
print(matrix)

print("2D arrays index: ")
print(matrix[0,0])
print(matrix[1,2])
print(matrix[2,1])
print("Negative indexing: ")
print(matrix[-1,-1])
print(matrix[-2,-2])
print(matrix[-3,-3])
```

```python
#3D array indexing
arr = np.array([
    [[1,2], [3,4]], #Block 0
    [[5,6], [7,8]] #Block 1
])
print(arr[0,0,0])
print(arr[0,1,1])
print(arr[1,0,1])
```

```python
#update array elements
arr = np.array([10,20,30,40])
arr[1]=96
print(arr)
```

```python
matrix = np.array([
    [1,2,3],
    [4,5,6]
])
matrix[1, 2] = 100
print(matrix)
```

```python
#index out of range
arr = np.array([10,20,30])
print(arr[5])
```

```python
marks = np.array([
    [85,90,78],
    [70,88,92],
    [95,81,96]
])

print("student 1, subject 1: ",marks[0,0])
print("student 2, subject 3: ",marks[1,2])
#Third student second subject marks update
marks[2,1]=86
print("update marks: ",marks)
```
