```python
import numpy as np

arr = np.array([
    [10,20,30],
    [40,50,60]
])
print(arr)
```

```python
#number of dimensions
a = np.array([10,20,30])

b = np.array([
    [10,20,30],
    [40,50,60]
])
c = np.array([
    [[1,2], [3,4]],
    [[5,6], [7,8]]
])

print("dimensions: ",a.ndim)
print("dimensions: ",b.ndim)
print("dimensions: ",c.ndim)
```

```python
#array structure
arr = np.array([
    [10,20,30],
    [40,50,60]
])
print(arr.shape)#2 Rows and 3 Columns
```

```python
arr = np.array([10,20,30,40])
print(arr.shape)
```

```python
#total elements
arr = np.array([
    [1,2,3],
    [4,5,6]
])
print(arr.size)
```

```python
#Data type
integers = np.array([10,20,30,40])
decimals = np.array([1.5,2.5,3.5,4.5])
names = np.array(["saurabh","Rohit","Aditya"])

print(integers.dtype)
print(decimals.dtype)
print(names.dtype)
```

```python
arr = np.array([10,20,30,40], dtype = np.int32)
print("Data type: ",arr.dtype)
print("Item size: ",arr.itemsize,"bytes")
```

```python
#find total arrays element

arr = np.array([10,20,30,40],dtype=np.int32)

print("Total elements: ",arr.size)
print("Each elements: ",arr.itemsize,"bytes")
print("Total data memory: ",arr.nbytes,"bytes")
```

```python
import numpy as np

arr = np.array([
    [10,20,30],
    [40,50,60]
],dtype=np.int32)

print("Arrays:\n",arr)
print("Dimensions:",arr.ndim)
print("Shape:",arr.shape)
print("total elements:",arr.size)
print("data type:",arr.dtype)
print("item size",arr.itemsize,"bytes")
print("total data memory:",arr.nbytes,"bytes")
```
