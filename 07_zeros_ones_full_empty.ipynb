```python
import numpy as np
```

```python
arr = np.zeros(5)
print(arr)
```

```python
arr = np.zeros(5, dtype=int)
print(arr)
```

```python
arr = np.zeros((2, 3), dtype=int)
print(arr)
#2 Rows and 3 Columns
```

```python
arr = np.ones(5, dtype=int)
print(arr)
```

```python
#3 Rows and 2 Columns
arr = np.ones((3, 2), dtype=int)
print(arr)
```

```python
arr = np.full(5, 7)
print(arr)
```

```python
#2D example
arr = np.full((2,3),100)
print(arr)
```

```python
arr = np.full(4, "Saurabh")
print(arr)
```

```python
arr = np.empty(5)
print(arr)
```

```python
#functions
import numpy as np

print("zeros:",np.zeros(4, dtype=int))
print("ones: ",np.ones(4, dtype=int))
print("full: ",np.full(4,9))
print("empty: ",np.empty(4))
```

```python
import numpy as np
student_marks = np.zeros(5, dtype=int)

print("initial marks: ",student_marks)

student_marks[0] = 85
student_marks[1] = 90
student_marks[2] = 97

print("update marks: ",student_marks)
```
