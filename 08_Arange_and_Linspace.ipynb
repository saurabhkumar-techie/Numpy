```python
import numpy as np
arr = np.arange(1, 10)
print(arr)
```

```python
arr = np.arange(2, 11, 2)
print(arr)
```

```python
#zero se sequence
arr = np.arange(5)
print(arr)
```

```python
#negative steps
arr = np.arange(10, 0, -2)
print(arr)
```

```python
#Decimal step
arr = np.arange(0, 1, 0.2)
print(arr)
```

```python
# #np.linspace()
# np.linspace(start, stop, num)
```

```python
arr = np.linspace(0, 10, 5)
print(arr)
```

```python
arr = np.linspace(1, 9, 3)
print(arr) # Three values
```

```python
#Ten values between 0 and 1
arr = np.linspace(0, 1, 10)
print(arr)
```

```python
#machine learning example
import numpy as np

learning_rates = np.linspace(0, 1, 5)
print("Learning rate values: ",learning_rates)
```

```python
even_numbers = np.arange(2, 101, 2)
print(even_numbers)
```

```python
#endpoint
arr = np.linspace(0, 10, 5, endpoint=False)
print(arr)
```
